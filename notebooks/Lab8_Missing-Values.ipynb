### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Write your code here
df.isna().sum(axis=0).reset_index(name="count").sort_values("count", ascending=False).head(30)
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
## Write your code here
sns.heatmap(df.isna()
	.sample(n=10, axis=1)
	.sample(frac=0.1, axis=0),
			yticklabels=False,
			cbar=False)
plt.title("Null Rows with Black Areas for Sampled 10 Columns", fontweight=700)
plt.xlabel("Columns");
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
## Write your code here
df.loc[df.YearsCode.isna()].shape[0]
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
## Write your code here
df = df.fillna({"YearsCode" :df.YearsCode.mode()[0]})
df.YearsCode.isna().sum()
```
