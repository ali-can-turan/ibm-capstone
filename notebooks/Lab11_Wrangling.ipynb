### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
min = df.ConvertedCompYearly.min()
max = df.ConvertedCompYearly.max()
df["ConvertedCompYearly_MinMax"] = (df.ConvertedCompYearly - min) / (max - min)
df["ConvertedCompYearly_MinMax"] 
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import math, numpy
df["ConvertedCompYearly_log"] = numpy.log1p(df.ConvertedCompYearly)
df["ConvertedCompYearly_log"]
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
df.YearsCodePro.value_counts()
df.YearsCodePro = df.YearsCodePro.replace({"More than 50 years": 50, "Less than 1 year": 1})
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df.YearsCodePro = df.YearsCodePro.astype("Int64")

bins = numpy.linspace(start=df.YearsCodePro.min(), stop=df.YearsCodePro.max(), num=5)
df["ExperienceLevel"] = pd.cut(x=df.YearsCodePro, bins=bins, labels=["junior", "knowledgable", "experienced", "pro"])
df.ExperienceLevel
```
