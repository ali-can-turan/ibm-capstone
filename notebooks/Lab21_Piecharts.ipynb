### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here
df["DatabaseList"] = df.DatabaseWantToWorkWith.str.split(";")
df_exploded = df.explode("DatabaseList")

df_group = (df_exploded
	.groupby("DatabaseList")
	.size()
	.reset_index(name="Count")
	.sort_values("Count", ascending=False)
	.head())
df_group

wedges, texts, autotexts = plt.pie(x=df_group.Count,
								   labels=df_group.DatabaseList,
								   autopct="%1.1f%%",
								   pctdistance=0.8,
								   wedgeprops=dict(width=0.4, lw=1, ec="1.0"),
                                   explode=[0.1, 0, 0, 0, 0],
                                   counterclock=False,
								   startangle=90);
for item in autotexts:
    item.set_color("white")
plt.title("Top 5 Database Popularity", fontweight=700);
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here
df_dummy = df.DevType.value_counts().reset_index().head()
df_dummy

wedges, texts, autotexts = plt.pie(x=df_dummy["count"],
								   labels=df_dummy.DevType,
								   autopct="%1.1f%%",
								   pctdistance=0.8,
								   wedgeprops=dict(width=0.4, lw=1, ec="1.0"),
                                   explode=[0.1, 0, 0, 0, 0],
                                   counterclock=False,
								   startangle=90)
for item in autotexts:
    item.set_color("white")

plt.title("Top 5 Developer Types", fontweight=700);
```
