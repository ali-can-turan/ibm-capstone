#### Step 4: Analysis of Experience and Job Satisfaction


```python
df.JobSat = df.JobSat.ffill().bfill()
df.RemoteWork = df.RemoteWork.ffill().bfill()
for col in ["JobSat", "RemoteWork"]:
    print(col, "-->", df[col].isna().sum())
```

Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here
df.YearsCodePro.isna().sum()
df.YearsCodePro = df.YearsCodePro.ffill().bfill()
df.YearsCodePro.isna().sum()

df.YearsCodePro = df.YearsCodePro.replace({"More than 50 years" : 50, "Less than 1 year" : 1})
df.YearsCodePro = pd.to_numeric(df.YearsCodePro, downcast="integer")
df.YearsCodePro.dtype

df.YearsCodePro = df.YearsCodePro.astype("Int64")
df.YearsCodePro.dtype
```

```python
import numpy as np

conditions = [
    (df.YearsCodePro < 6),
    (df.YearsCodePro >= 6) & (df.YearsCodePro < 10),
    (df.YearsCodePro >= 10) & (df.YearsCodePro < 20),
    (df.YearsCodePro >= 20)
]
choices = ["0-5", "5-10", "10-20", ">20"]

df["YearsCodePro.cat"] = np.select(conditions, choices, default="diğer")
```

#### Step 5: Visualize Job Satisfaction


```python
df["YearsCodePro.cat"] = pd.Categorical(df["YearsCodePro.cat"], categories = ["0-5", "5-10", "10-20", ">20"], ordered=True)
df_summarized = df.groupby("YearsCodePro.cat")["JobSat"].median()
df_summarized
```

```python
df_dummy = df.groupby("YearsCodePro.cat").size()
df_dummy
```

```python
plt.style.use("ggplot")
plt.rc("axes.spines", top=False, right=False)

ax = df_summarized.plot.bar(xlabel="Years in Code", ylabel="Satisfaction", color="orange")
plt.title(label="Job satisfaction by Years in Code", fontweight="bold", y=1.04)
plt.xticks(rotation=0)
fig = ax.get_figure()
fig.set_layout_engine("constrained")
fig.savefig("JobSat.WorkExp.png");
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here
# transform the column in question into list to explode and explode
df_exploded = (df
	.assign(Language=df["LanguageHaveWorkedWith"].str.split(';'))
	.explode("Language"))

# aggregate
counts = df_exploded.groupby(['Country', 'Language']).size().reset_index(name='Count')

# Find the index of max-count language for each country
idx = counts.groupby('Country')['Count'].idxmax()

# feed the counts df with those indexes to find the most popular language by country
top_languages = counts.loc[idx].reset_index(drop=True)

# pivot the table to feed heatmap
counts_pivot = counts.pivot(index="Country", columns="Language", values="Count")
sns.heatmap(data=counts_pivot.sample(10),
			center=200,
			square=False,
			ec="0.0",
			lw=1,
			cmap="viridis",
			vmax=500,
			alpha=0.8)
plt.title("Country / Language Count", y=1.05)
plt.xticks(rotation=90);
```
