```python
import requests # you need this module to make an API call
import pandas as pd
```

```python
api_url = "http://api.open-notify.org/astros.json" # this url gives use the astronaut data
```

```python
response = requests.get(api_url) # Call the API using the get method and store the
                                # output of the API call in a variable called response.
```

```python
if response.ok:             # if all is well() no errors, no network timeouts)
    data = response.json()  # store the result in json format in a variable called data
                            # the variable data is of type dictionary.
```

```python
print(data)   # print the data just to check the output or for debugging
```

Print the number of astronauts currently on ISS.


```python
print(data.get('number'))
```

Print the names of the astronauts currently on ISS.


```python
astronauts = data.get('people')
print("There are {} astronauts on ISS".format(len(astronauts)))
print("And their names are :")
for astronaut in astronauts:
    print(astronaut.get('name'))
```

#### Write a function to get the number of jobs for the Python technology.

 ##### The keys in the json are

You can also view  the json file contents  from the following URL.

https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json



```python
api_url="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"
response = requests.get(api_url).json()
```

```python
def get_number_of_jobs_T(technology):
    #your code goes here
    number_of_jobs = 0
    for item in response:
        if technology in item.get("Key Skills"):
            number_of_jobs += 1
    return technology,number_of_jobs
```

Calling the function for Python and checking if it works.


```python
get_number_of_jobs_T("Python")
```

#### Write a function to find number of jobs in US for a location of your choice


```python
 def get_number_of_jobs_L(location):
    #your coe goes here
    number_of_jobs = 0
    response = requests.get(api_url).json()
    for item in response:
        if location in item.get("Location"):
            number_of_jobs += 1
    return location,number_of_jobs
```

Create a python list of all locations for which you need to find the number of jobs postings.


```python
#your code goes here
location_list = []
response = requests.get(api_url).json()
for item in response:
    location = item.get("Location")
    location_list.append(location)
print(len(location_list))
```

Import libraries required to create excel spreadsheet


```python
# your code goes here
# !pip install openpyxl
from openpyxl import Workbook
```

Create a workbook and select the active worksheet


```python
# your code goes here
wb=Workbook()
ws=wb.active 
```

Find the number of jobs postings for each of the location in the above list.
Write the location name and the number of jobs postings into the excel spreadsheet.


```python
#your code goes here
ws.append(['location','number_of_jobs'])
location_set = set(location_list)
for item in location_set:
    ws.append([item, location_list.count(item)])
```

Save into an excel spreadsheet named **job-postings.xlsx**.


```python
#your code goes here
wb.save("job-postings.xlsx")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt 
plt.rc("axes.spines", top=False, right=False)
sns.set_style("darkgrid")
df=pd.read_excel("job-postings.xlsx")

fig, ax = plt.subplots(figsize=(6, 4), layout="constrained")
sns.barplot(data=df.sort_values("number_of_jobs", ascending=False),
			x="location",
			y="number_of_jobs",
			hue="location",
			palette="husl",
			ax=ax)
plt.xticks(rotation=90)
plt.ylabel("Jobs Quantity")
plt.xlabel("Location")
plt.title("Number of Jobs by Location", fontweight=700, y=1.03)

fig.savefig("jobs.png")
```
